```python
import pandas as pd
import matplotlib.pyplot as plt

history = pd.read_csv("../outputs/training_history.csv")

plt.figure(figsize=(8,5))
plt.plot(history["accuracy"], label="Training")
plt.plot(history["val_accuracy"], label="Validation")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
```
